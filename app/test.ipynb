```python
import os
import joblib
from functools import lru_cache
```

```python
APP_DIR_NAME='app'
MODEL_DIR_NAME='models'
MODEL_NAME='pipeline.joblib'
```

```python
model_path=os.path.join(APP_DIR_NAME,MODEL_DIR_NAME,MODEL_NAME)
```

```python
model_path
```

```python
@lru_cache(maxsize=1)
def get_model():
    model = joblib.load(model_path)
    return model
```
